# Week 3: Viewing HREA lightscore rasters (Google Colab)

Opens HREA lightscore GeoTIFFs for Ghana and Burundi and renders them on latitude/longitude axes.

Locally these plots came out black. Here the colorbar shows why: nodata cells are stored as roughly -1,000,000, which stretches the color scale so far that real values are hard to see. Later weeks mask the nodata value before analysis.

Data is read from Google Drive; update `DATA_DIR` to your own folder.

```python
import matplotlib.pyplot as plt
import rasterio

DATA_DIR = "/content/drive/My Drive/Energy and Health/Data/Raw/LEN"
file_path = f"{DATA_DIR}/GHA/GHA_set_lightscore_2013.tif"

with rasterio.open(file_path) as dataset:
    data = dataset.read(1)
    bounds = dataset.bounds

plt.figure(figsize=(10, 6))
plt.imshow(data, cmap="viridis", origin="upper",
           extent=[bounds.left, bounds.right, bounds.bottom, bounds.top])
plt.colorbar(label="Light Score Intensity")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Ghana Light Score (2013)")
plt.show()
```
